Question 2: What are the key assumptions of Simple Linear Regression?
Ans:- The key assumptions of Simple Linear Regression (SLR) ensure that the model produces reliable and unbiased results. These assumptions are:

1. Linearity

There is a linear relationship between the independent variable (X) and the dependent variable (Y).

This means Y changes at a constant rate with respect to X.

Can be checked using a scatter plot.

2. Independence of Errors

The residuals (errors) are independent of each other.

No correlation between consecutive errors.

Important especially for time-series data.

3. Homoscedasticity

The residuals have constant variance across all levels of X.

Errors should be spread evenly around the regression line.

Unequal spread indicates heteroscedasticity.

4.  Normality of Errors

The residuals are normally distributed.

This is mainly required for hypothesis testing and confidence intervals.

Checked using histograms or Q–Q plots.

5. No Multicollinearity (Automatically satisfied in SLR)

Since SLR has only one independent variable, multicollinearity is not an issue.

This assumption becomes important in multiple linear regression.

6. Zero Mean of Errors

The average of the error terms is zero, meaning the model is unbiased.

Question 5: What is the method of least squares in linear regression?

Ans:-
Method of Least Squares in Linear Regression

The method of least squares is a mathematical technique used to estimate the parameters of a linear regression model by finding the best-fitting line through the data points.

Core Idea

It chooses the regression line such that the sum of the squared residuals (errors) is minimized.

- Residual = Actual value − Predicted value

- Squaring ensures:

- All errors are positive

- Larger errors are penalized more heavily

Purpose of Least Squares

- Finds the best-fit line

- Minimizes prediction error

- Provides unbiased and efficient estimates (under SLR assumptions)

- Makes predictions as close as possible to actual data

Question 6: What is Logistic Regression? How does it differ from Linear Regression?

Ans:- Logistic Regression is a supervised machine learning algorithm used for classification problems, especially binary classification (e.g., Yes/No, 0/1, Pass/Fail).

Instead of predicting a continuous value, it predicts the probability that an input belongs to a particular class.

| ---------------------- | ------------------------- | ------------------------------- |
| **Type of Problem**    | Regression                | Classification                  |
| **Output**             | Continuous values         | Probability (0–1)               |
| **Dependent Variable** | Numerical                 | Categorical (binary)            |
| **Model Equation**     | (Y = \beta_0 + \beta_1 X) | (P(Y=1) = \frac{1}{1 + e^{-z}}) |

Question 7: Name and briefly describe three common evaluation metrics for regression models?

Regression metrics measure how well a model’s predictions match the actual values.
Three commonly used evaluation metrics are:
1. Mean Absolute Error (MAE)

Definition:
MAE is the average of the absolute differences between actual and predicted values.

MAE =1/𝑛∑∣𝑌𝑖−𝑌^i|

2. Mean Squared Error (MSE)

3. Root Mean Squared Error (RMSE)

Question 8: What is the purpose of the R-squared metric in regression analysis?

Ans:- Purpose of the R-squared (R²) Metric in Regression Analysis

R-squared (R²) is a statistical metric used to measure how well a regression model explains the variability of the dependent variable.

Definition

R-squared represents the proportion of the total variation in the target variable (Y) that is explained by the independent variable(s) (X).

Purpose of R-squared

1. Measure Goodness of Fit
Indicates how well the regression model fits the data.

2. Explain Variability
Shows the percentage of variation in Y explained by X.

- Example: R² = 0.85 → 85% of the variance in Y is explained by the model.

3. Model Comparison
Helps compare regression models with the same dependent variable.

4. Model Interpretation
Provides an intuitive understanding of model performance.

```python
# Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.

# Import required libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Independent variable
y = np.array([2, 4, 5, 4, 5])                 # Dependent variable

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Print slope and intercept
print("Slope (β1):", model.coef_[0])
print("Intercept (β0):", model.intercept_)
```
